```python
print(df.info())
print(df.shape)
print(df.columns)
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()

print(df.columns)
```

```python
# Convert Call Duration into Seconds
def time_to_seconds(time):
    h, m, s = map(int, time.split(":"))
    return h * 3600 + m * 60 + s

df["Duration_Seconds"] = df["Call Duration"].apply(time_to_seconds)
```

#### Question 1 – Find Distinct From Numbers

#### Question 3 – Total Call Duration Per User

```python
duration = df.groupby("From Mobile")["Duration_Seconds"].sum()

print(duration)
```

```python
def convert(sec):
    h = sec // 3600
    m = (sec % 3600) // 60
    s = sec % 60
    return f"{h:02}:{m:02}:{s:02}"

duration = duration.apply(convert)

print(duration)
```

#### Question 4 – Calculate Income

```python
income = df[df["Duration_Seconds"] >= 60]["Duration_Seconds"].apply(
    lambda x: (x / 60) * 0.30
).sum()

print(f"Total Income = ₹{income:.2f}")
```

```python
income = df[df["Duration_Seconds"] >= 60]["Duration_Seconds"].apply(
    lambda x: math.ceil(x / 60) * 0.30
).sum()
```
