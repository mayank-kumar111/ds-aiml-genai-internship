# **Q12. What is the difference between correlation and causation? Give an example. (3 Marks)**

### **Answer:**

**Correlation:**
Correlation measures the strength and direction of the relationship between two variables. It does not imply that one variable causes the other.

**Causation:**
Causation means that a change in one variable directly causes a change in another variable.

**Example:**

* **Correlation:** Ice cream sales and drowning incidents increase during summer. They are correlated because both increase due to hot weather.
* **Causation:** Pressing the accelerator of a car increases its speed. The accelerator directly causes the speed to increase.

# **Q13. Explain the terms Bias and Variance. What is the bias–variance trade-off? (3 Marks)**

### **Answer:**

**Bias:**
Bias is the error caused by overly simple assumptions in the learning algorithm. High bias leads to **underfitting**.

**Variance:**
Variance is the error caused by excessive sensitivity to training data. High variance leads to **overfitting**.

**Bias–Variance Trade-off:**
The bias–variance trade-off is the balance between bias and variance to achieve the best model performance. A good model should have both low bias and low variance.

# **Q19. Explain Simple Linear Regression. Derive the cost function (MSE) and describe how the parameters (slope and intercept) are estimated. Illustrate with the line equation (y = mx + c). (8 Marks)** 

## **Answer**

### Definition

Simple Linear Regression is a supervised machine learning algorithm used to predict a continuous dependent variable based on one independent variable. It models the relationship between two variables by fitting the best straight line.

### Equation of the Line

[
y = mx + c
]

Where:

* (y) = Predicted output (dependent variable)
* (x) = Input (independent variable)
* (m) = Slope (weight)
* (c) = Intercept (bias)

### Cost Function (Mean Squared Error)

The Mean Squared Error (MSE) measures the average squared difference between actual and predicted values.



### Parameter Estimation

The parameters (m) (slope) and (c) (intercept) are estimated by minimizing the MSE using optimization techniques such as Gradient Descent or analytical methods.

### Applications

* House price prediction
* Salary prediction
* Sales forecasting
* Temperature prediction

# **Q21. Explain the Gradient Descent optimization algorithm. Contrast Batch, Stochastic, and Mini-batch Gradient Descent, listing one advantage and one disadvantage of each. Discuss the role of the learning rate. (8 Marks)** 

## **Answer**

### Definition

Gradient Descent is an optimization algorithm used to minimize the cost function by updating model parameters in the direction of the negative gradient.



### Types of Gradient Descent

| Type                              | Advantage                    | Disadvantage                        |
| --------------------------------- | ---------------------------- | ----------------------------------- |
| Batch Gradient Descent            | Stable convergence           | Slow for large datasets             |
| Stochastic Gradient Descent (SGD) | Faster updates               | Noisy convergence                   |
| Mini-batch Gradient Descent       | Balances speed and stability | Batch size must be chosen carefully |

### Learning Rate

The learning rate controls the step size during parameter updates.

* Too high: May overshoot the minimum.
* Too low: Training becomes very slow.
* Proper learning rate: Faster and stable convergence.

A confusion matrix is a table used to evaluate the performance of a classification model.

|                 | Predicted Positive | Predicted Negative |
| --------------- | ------------------ | ------------------ |
| Actual Positive | TP                 | FN                 |
| Actual Negative | FP                 | TN                 |

### Definitions

* **TP (True Positive):** Correctly predicted positive.
* **TN (True Negative):** Correctly predicted negative.
* **FP (False Positive):** Predicted positive but actually negative.
* **FN (False Negative):** Predicted negative but actually positive.

### Formulas



### Why Accuracy Can Be Misleading

In imbalanced datasets, one class may dominate. A model can achieve high accuracy by predicting only the majority class while performing poorly on the minority class. Therefore, Precision, Recall, and F1-Score provide a better evaluation.

# **Q23. Discuss common sampling techniques in machine learning—simple random, stratified, systematic, and cluster sampling. State when each is appropriate. (8 Marks)** 

## **Answer**

### 1. Simple Random Sampling

Every sample has an equal chance of being selected.

**Appropriate when:** The population is homogeneous.

### 2. Stratified Sampling

The population is divided into strata, and samples are selected proportionally from each group.

**Appropriate when:** The dataset is imbalanced and class proportions need to be preserved.

### 3. Systematic Sampling

Every (k^{th}) sample is selected after a random starting point.

**Appropriate when:** The data is ordered and evenly distributed.

### 4. Cluster Sampling

The population is divided into clusters, and entire clusters are randomly selected.

**Appropriate when:** The population is geographically dispersed or naturally grouped.

# **Q25. A binary classifier produces the following results on 100 samples:**

* **TP = 40**
* **TN = 30**
* **FP = 20**
* **FN = 10**

**Compute Accuracy, Precision, Recall, and F1-Score.** 

## **Answer**

### Given

* TP = 40
* TN = 30
* FP = 20
* FN = 10

### 1. Accuracy

**Formula**

[
Accuracy=\frac{TP+TN}{TP+TN+FP+FN}
]

**Calculation**

[
=\frac{40+30}{40+30+20+10}
]

[
=\frac{70}{100}
]

[
=0.70
]

**Answer**

[
\boxed{Accuracy=70%}
]

---

### 2. Precision

**Formula**

[
Precision=\frac{TP}{TP+FP}
]

**Calculation**

[
=\frac{40}{40+20}
]

[
=\frac{40}{60}
]

[
=0.6667
]

**Answer**

[
\boxed{Precision=66.67%}
]

---

### 3. Recall

**Formula**

[
Recall=\frac{TP}{TP+FN}
]

**Calculation**

[
=\frac{40}{40+10}
]

[
=\frac{40}{50}
]

[
=0.80
]

**Answer**

[
\boxed{Recall=80%}
]

---

### 4. F1-Score

**Formula**

[
F1=\frac{2\times Precision\times Recall}{Precision+Recall}
]

**Calculation**

[
=\frac{2\times0.6667\times0.80}{0.6667+0.80}
]

[
=\frac{1.0667}{1.4667}
]

[
=0.7273
]

**Answer**

[
\boxed{F1\text{-}Score=72.73%}
]

Perform **ONE step of SGD** for linear regression (prediction (\hat y = wx + b), squared-error loss) and report the updated **w** and **b**. 

## **Answer**

### Step 1: Prediction

[
\hat y=wx+b
]

[
=(0.5)(2)+0
]

[
=1
]

---

### Step 2: Error

[
Error=\hat y-y
]

[
=1-3
]

[
=-2
]

---

### Step 3: Gradient for Weight

[
\frac{\partial J}{\partial w}=2(Error)x
]

[
=2(-2)(2)
]

[
=-8
]

---

### Step 4: Gradient for Bias

[
\frac{\partial J}{\partial b}=2(Error)
]

[
=2(-2)
]

[
=-4
]

---

### Step 5: Update Weight

[
w_{new}=w-\alpha\frac{\partial J}{\partial w}
]

[
=0.5-0.1(-8)
]

[
=0.5+0.8
]

[
=1.3
]

---

### Step 6: Update Bias

[
b_{new}=b-\alpha\frac{\partial J}{\partial b}
]

[
=0-0.1(-4)

[
=0.4
