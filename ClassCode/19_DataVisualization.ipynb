```python
dates = pd.date_range("24/06/2026 8:00am", periods=6, freq='2H')
dates
```

```python
temp = [27.6, 31.5, 35.6, 38.7, 37.3, 33.2]
```

```python
plt.figure(figsize=(12, 8), dpi=140)
plt.plot(dates, temp)
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp)
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp)
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='k')
plt.plot(dates, temp)
plt.xlabel("Dates", size=15, color='w')
plt.ylabel("Temp", size=15, color='w')
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.tick_params(color='w', labelcolor='w')
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp, 'r--o')
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp, 'r--o', mfc='k', mec='y')
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp, 'r--o', mfc='k', mec='y', label='temp')
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.grid()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp, 'r--o', mfc='k', mec='y', label='temp')
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.grid()
plt.legend()
plt.savefig('fig.jpeg')
plt.show()
```
