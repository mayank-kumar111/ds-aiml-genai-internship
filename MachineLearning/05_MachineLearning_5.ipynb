    document  -->   single row of feature
    corpus    -->   collection of document
    vocab     -->   collection of unique words in corpus

```python
corpus = [
    'hello hii bye',                # d1
    'python is awesome',            # d2
    'hii bye hii hello',            # d3
    'python is easy to learn'       # d4
]
```

```python
corpus
```

```python
vocab = []

for doc in corpus:
    for i in doc.split():
        if i not in vocab:
            vocab.append(i)
else:
    print(vocab)
```

```python
df = pd.read_csv('spam.csv', encoding='latin')
df
```

```python
spam_ham = df[['v2', 'v1']]
spam_ham
```

```python
spam_ham.rename(columns={'v2':'Documents', 'v1':'label'}, inplace=True)
```

```python
spam_ham
```

```python
X = spam_ham.drop('label', axis=1)
X
```

```python
y = spam_ham.label
y
```

```python
corpus = X['Documents'].values
corpus
```

```python
len(corpus)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
count_vec = CountVectorizer()
count_vec
```

```python
count_vec.fit(corpus)
```

```python
count_vec.vocabulary_
```

```python
corpus_tr = count_vec.transform(corpus).toarray()
corpus_tr
```

```python
y
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import SGDClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier


from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
log = LogisticRegression()
log.fit(corpus_tr, y)
y_pred = log.predict(corpus_tr)

print(f"Accuracy_score: {accuracy_score(y, y_pred)}")
print(f"Confusion_matrix: {confusion_matrix(y, y_pred)}")
print(f"Classification report: {classification_report(y, y_pred)}")
```
