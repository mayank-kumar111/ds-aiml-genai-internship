```python
temp = [27.6, 31.5, 35.6, 38.7, 37.3, 33.2]
```

```python
plt.figure(figsize=(12, 8), dpi=140)
plt.plot(dates, temp) # Plot the temperature data against the dates
plt.xlabel("Dates", size=15) # Set the label for the x-axis
plt.ylabel("Temp", size=15) # Set the label for the y-axis
plt.title("Time VS Temp", size=15) # Set the title of the plot
plt.show() # Display the plot
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='lightgrey')
plt.plot(dates, temp)
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp)
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90) # Rotate the x-axis tick labels by 90 degrees for better readability
plt.yticks(range(25, 41)) # Set the y-axis tick labels to range from 25 to 40
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='k')
plt.plot(dates, temp)
plt.xlabel("Dates", size=15, color='w')
plt.ylabel("Temp", size=15, color='w')
plt.title("Time VS Temp", size=15, color='w')
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.tick_params(color='w', labelcolor='w') # Set the color of the ticks and tick labels to white
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp, 'r--o')
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='lightgrey')
plt.plot(dates, temp, 'o', mfc='k', mec='g', ms=10) # mfc = marker face color, mec = marker edge color, ms = marker size
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.show()
```

```python
plt.figure(figsize=(12, 8), dpi=140, facecolor='c')
plt.plot(dates, temp, 'r--o', mfc='k', mec='y', label='temp')
plt.xlabel("Dates", size=15)
plt.ylabel("Temp", size=15)
plt.title("Time VS Temp", size=15)
plt.xticks(rotation=90)
plt.yticks(range(25, 41))
plt.grid() # Add grid lines to the plot
plt.legend() # Add a legend to the plot, legend() function is used to display the legend on the plot, which shows the label for each plotted line or marker. The label is specified in the plot() function using the label parameter.
plt.show()
```

#### Bar Plot
1. Bar plot is used to represent categorical data with rectangular bars, where the length of each bar is proportional to the value it represents. It is commonly used to compare different categories or groups within a dataset. In this case, we are using a bar plot to visualize the total bill amounts for different days of the week in the "tips" dataset.

```python
tips = sns.load_dataset("tips")
taxi = sns.load_dataset("taxis")
```

```python
tips
```

```python
taxi
```

```python
tips['sex'].value_counts()
```

```python
tips['sex'].value_counts().plot(kind='bar') # plot kind='bar' is used to create a bar plot of the value counts
```

```python
i1 = tips['sex'].value_counts().index
i1
```

```python
v1 = tips['sex'].value_counts().values
v1
```

```python
plt.bar( i1, v1) # Create a bar plot with i1 as the x-axis values and v1 as the y-axis values
```

```python
plt.bar( i1, v1, color=['r', 'g']) # Create a bar plot with i1 as the x-axis values and v1 as the y-axis values, and set the color of the bars to red and green respectively
```

```python
i2 = tips['smoker'].value_counts().index
v2 = tips['smoker'].value_counts().values
```

```python
i2
```

```python
v2
```

```python
plt.bar(i1, v1, color=['r', 'g', 'c'], alpha=0.3) # Create a bar plot with i1 as the x-axis values and v1 as the y-axis values, set the color of the bars to red, green, and cyan respectively, and set the transparency of the bars to 0.3 (30% opaque)
plt.bar(i2, v2, color=['r', 'g', 'c']) # Create a bar plot with i2 as the x-axis values and v2 as the y-axis values, set the color of the bars to red, green, and cyan respectively
plt.show()
```

#### Pie chart
1. Pie Chart is use to visualize the proportion of different categories in a dataset. It is a circular chart divided into sectors, where each sector represents a category and its size corresponds to the proportion of that category in the dataset. Pie charts are useful for showing relative sizes of parts to a whole, but they can be less effective when there are many categories or when precise comparisons are needed.

```python
i1
```

```python
v1
```

```python
plt.pie(v1, labels=i1) # Create a pie chart with v1 as the values and i1 as the labels for each slice of the pie chart
plt.show()
```

```python
plt.pie(v1, labels=i1, autopct="%.2f%%") # Display the percentage of each slice in the pie chart with two decimal places
plt.show()
```

```python
plt.pie(v1, labels=i1, autopct="%.2f%%", explode=[0.3,0], shadow=True) # Create a pie chart with v1 as the values and i1 as the labels for each slice of the pie chart, display the percentage of each slice with two decimal places, explode the first slice by 0.3, and add a shadow effect to the pie chart
plt.show()
```

#### Scatter plot
Scatter plot is a type of data visualization that displays the relationship between two numerical variables. Each point on the scatter plot represents an observation in the dataset, with its position determined by the values of the two variables being compared. Scatter plots are useful for identifying patterns, trends, correlations, and potential outliers in the data.


```python
tips
```

```python
plt.scatter(tips['total_bill'], tips['tip'])
plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'])
plt.show()
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue=tips['sex']) # hue parameter is used to color the points based on the values
plt.show()
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue=tips['time'])
plt.show()
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue=tips['day'])
plt.show()
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue=tips['sex'], size=tips['size'])
plt.show()
```

#### Histogram

Histogram is a graphical representation of the distribution of a dataset. It is created by dividing the range of the data into intervals (bins) and counting the number of observations that fall into each bin. The resulting plot consists of adjacent rectangles (bars) where the height of each bar represents the frequency (count) of observations in that bin. Univarant Graph only plot single value.

```python
tips['total_bill']
```

```python
plt.hist(tips['total_bill'], bins=25, ec='y') # ec='y' parameter is used to set the edge color of the bars to yellow and bins=25 parameter is used to set the number of bins to 25.
plt.show()
```

```python
sns.histplot(x=tips['total_bill'])
plt.show()
```

##### Subplots

Subplots is used to create multiple plots in a single figure. It allows you to arrange multiple plots in a grid-like layout, making it easier to compare and visualize different datasets or aspects of the same dataset.

```python
taxi
```

```python
plt.figure(figsize=(12,8), dpi=150, facecolor='c')

plt.subplot(2,2,1)
sns.scatterplot(x=tips['total_bill'], y=tips['tip'])

plt.subplot(2,2,2)
sns.histplot(x=taxi['fare'])

plt.subplot(2,2,3)
taxi['payment'].value_counts().plot(kind='bar') # plot kind='bar' is used to create a bar plot of the value counts

plt.subplot(2,2,4)
taxi['color'].value_counts().plot(kind='bar')
```
