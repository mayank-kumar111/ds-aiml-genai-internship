### Stratified sampling

```python
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
s = StratifiedShuffleSplit(n_splits=10, test_size=0.2, random_state=121)
s
```

```python
obj = s.split(X, X['ocean_proximity'])
obj
```

```python
for i in obj:
    print(i)
    print()
```

```python
c = 1
for i in s.split(X, X['ocean_proximity']):
    print(f"Split: {c}")
    print(i)
    print()
    c += 1
```

```python
c = 1
for i,j in s.split(X, X['ocean_proximity']):
    print(f"Split: {c}")
    print(f"Train_idx: {i}")
    print(f"Test_idx: {j}")
    print()
    c += 1
```

```python
s = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=121)
obj = s.split(X, X['ocean_proximity'])
obj
```
