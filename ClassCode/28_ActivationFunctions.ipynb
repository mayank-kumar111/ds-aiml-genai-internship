#  What is an Activation Function?
An activation function is a mathematical function applied to the output of a neuron.
- Without it, a neural network would just be a linear model (like linear regression).
- With it, the network can learn complex, non-linear patterns (like images, language, speech).

# Why Do We Need Activation Functions?
1. Add Non-Linearity

- Real-world data is rarely linear.
- Without activation, the network can’t capture complex patterns.
- Ex: To recognize a cat in an image, the relationship between pixels and the label “cat” is non-linear.

2. Control Output Range

- Some activation functions squash outputs to a certain range (like 0–1 or -1–1).
- This keeps the model stable and interpretable.

3. Enable Deep Networks

- Without activation, stacking multiple layers would collapse into a single linear function.
- With activation, each layer adds new levels of abstraction.
- Example:
    - 1st layer learns edges
    - 2nd layer learns shapes
    - 3rd layer learns objects

# Common Activation Functions in Neural Networks

| **Activation Function**           | **Formula**                                                            | **Output Range**     | **Use Case**                               | **Advantages**                                             | **Disadvantages**                                    |
| --------------------------------- | ---------------------------------------------------------------------- | -------------------- | ------------------------------------------ | ---------------------------------------------------------- | ---------------------------------------------------- |
| **Sigmoid**                       | $f(z) = \frac{1}{1+e^{-z}}$                                            | (0, 1)               | Probabilities in binary classification     | Smooth output, interpretable as probability                | Vanishing gradients, outputs not zero-centered       |
| **ReLU (Rectified Linear Unit)**  | $f(z) = \max(0, z)$                                                    | \[0, ∞)              | Most common in hidden layers for deep nets | Simple, fast, avoids vanishing gradient, efficient         | Dying ReLU problem (neurons stuck at 0)              |
| **Leaky ReLU**                    | $f(z) = \max(0.01z, z)$                                                | (-∞, ∞)              | Alternative to ReLU                        | Solves dying ReLU problem, allows small gradients when z<0 | Small negative slope may not always help             |
| **Softmax**                       | $f(z_i) = \frac{e^{z_i}}{\sum_j e^{z_j}}$                              | (0, 1) and sums to 1 | Multi-class classification output layer    | Converts scores to probabilities, interpretable            | Computationally expensive, sensitive to large values |


- Sigmoid / Tanh → Older, used in small networks (good for probabilities).
- ReLU → Standard choice for hidden layers in deep learning.
- Leaky ReLU / ELU / Swish → Improvements over ReLU.
- Softmax → Always used in the output layer for multi-class classification.

# Master Table of Activation Functions

| **Activation Function**           | **Formula**                                                                          | **Output Range** | **Use Case**                      | **Advantages**                                | **Disadvantages**                                              |
| --------------------------------- | ------------------------------------------------------------------------------------ | ---------------- | --------------------------------- | --------------------------------------------- | -------------------------------------------------------------- |
| **Step Function**                 | $f(z) = \begin{cases}1 & z \geq 0 \\ 0 & z < 0 \end{cases}$                          | {0,1}            | Very simple binary classification | Easy to implement, clear threshold            | Not differentiable, no gradient → not useful for deep learning |
| **Linear (Identity)**             | $f(z) = z$                                                                           | (-∞, ∞)          | Regression problems               | Simple, keeps values proportional             | No non-linearity → useless in deep networks                    |
| **Sigmoid (Logistic)**            | $f(z) = \frac{1}{1+e^{-z}}$                                                          | (0, 1)           | Binary classification output      | Probabilistic interpretation                  | Vanishing gradient, not zero-centered                          |
| **Tanh (Hyperbolic Tangent)**     | $f(z) = \tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$                                | (-1, 1)          | Hidden layers (before ReLU era)   | Zero-centered, stronger gradient than sigmoid | Vanishing gradient                                             |
| **ReLU (Rectified Linear Unit)**  | $f(z) = \max(0, z)$                                                                  | \[0, ∞)          | Hidden layers in deep nets        | Fast, efficient, avoids vanishing gradient    | Dying ReLU problem                                             |
| **Leaky ReLU**                    | $f(z) = \max(\alpha z, z), \ \alpha \approx 0.01$                                    | (-∞, ∞)          | Alternative to ReLU               | Fixes dying ReLU, small slope when z<0        | α must be tuned, not always helpful                            |
| **Parametric ReLU (PReLU)**       | Same as Leaky ReLU but α is trainable                                                | (-∞, ∞)          | Advanced networks                 | Learns slope automatically                    | More parameters, risk of overfitting                           |
| **Randomized ReLU (RReLU)**       | Negative slope is random during training                                             | (-∞, ∞)          | Regularization in deep nets       | Adds randomness, reduces overfitting          | Inference behavior differs                                     |
| **ELU (Exponential Linear Unit)** | $f(z) = \begin{cases} z & z>0 \\ \alpha(e^z-1) & z \leq 0 \end{cases}$               | (-α, ∞)          | Deep nets with faster learning    | Allows negative outputs, continuous           | More expensive than ReLU                                       |
| **SELU (Scaled ELU)**             | $f(z) = \lambda \cdot \begin{cases} z & z>0 \\ \alpha(e^z-1) & z \leq 0 \end{cases}$ | (-∞, ∞)          | Self-normalizing networks         | Keeps mean & variance stable                  | Works best with specific initialization                        |
| **Swish (Google)**                | $f(z) = z \cdot \sigma(z)$                                                           | (-∞, ∞)          | Modern deep networks              | Smooth, non-monotonic, often better than ReLU | Slower than ReLU                                               |
| **Mish**                          | $f(z) = z \cdot \tanh(\ln(1+e^z))$                                                   | (-∞, ∞)          | Alternative to Swish/ReLU         | Smooth, good gradient flow                    | Computationally heavy                                          |
| **Maxout**                        | $f(z) = \max(w_1x+b_1, w_2x+b_2)$                                                    | (-∞, ∞)          | Robust deep nets                  | Learns best activation shape                  | More parameters, heavy computation                             |
| **Softmax**                       | $f(z_i) = \frac{e^{z_i}}{\sum_j e^{z_j}}$                                            | (0,1), sums to 1 | Multi-class classification        | Outputs probabilities, interpretable          | Sensitive to large input values                                |


- Old activations → Step, Linear, Sigmoid, Tanh (rarely used in hidden layers now).
- Modern standards → ReLU (default), Leaky ReLU, ELU, SELU.
- Newer research → Swish, Mish (sometimes outperform ReLU).
- Output functions → Sigmoid (binary), Softmax (multi-class), Linear (regression).
