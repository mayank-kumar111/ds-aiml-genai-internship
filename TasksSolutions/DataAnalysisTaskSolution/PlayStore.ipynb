#### Analyse play store dataset and write down conclusion of the analysis and also create plots if possible.

```python
import pandas as pd
data = pd.read_csv('D:\\Program\\Data_Science\\Data\\GooglePlayStore.csv')
import matplotlib.pyplot as plt
data
```

## que1 :- Which category of apps has big count ?

```python
cat = data['Category'].value_counts()
print(cat)
cat.plot(kind='bar')
plt.xlabel('Category')
plt.ylabel('Count')
plt.title('Distribution of Apps by Category')
plt.show()
```

## que2 :- Top 5 Highest rating app from each category?

```python
top_5_apps = data.groupby('Category').apply(lambda x: x.nlargest(5, 'Rating'))
top_5_apps
```

```python
top_5_apps = data.groupby('Category').apply(lambda x: x.nlargest(5, 'Rating')).reset_index(drop=True)
top_5_apps
```

## que3 :- Find number of free and paid apps.

```python
free_apps = data[data['Type'] == 'Free']
paid_apps = data[data['Type'] == 'Paid']

print(f"Number of free apps: {len(free_apps)}")
print(f"Number of paid apps: {len(paid_apps)}")

plt.figure(figsize=(6, 4))
plt.bar(['Free', 'Paid'], [len(free_apps), len(paid_apps)])
plt.xlabel('App Type')
plt.ylabel('Count')
plt.title('Distribution of Free and Paid Apps')
plt.show()
```

## que4 :- Number of highest rating apps?

```python
highest_rated_apps = data[data['Rating'] == data['Rating'].max()]
print(f"Number of highest rating apps: {len(highest_rated_apps)}")
```
